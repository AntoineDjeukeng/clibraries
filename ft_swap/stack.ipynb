```python
from collections import deque
import random

class Stack:
    def __init__(self, values):
        self.a = deque(values)
        self.b = deque()
        self.ops = []
        self.size = len(values)
        self.indexed = self.indexify(values)

    def indexify(self, values):
        sorted_vals = sorted(set(values))
        return {val: i for i, val in enumerate(sorted_vals)}

    def push(self, src, dst, name):
        if src:
            dst.appendleft(src.popleft())
            self.ops.append(name)

    def swap(self, s, name):
        if len(s) >= 2:
            s[0], s[1] = s[1], s[0]
            self.ops.append(name)

    def rotate(self, s, name):
        if s:
            s.append(s.popleft())
            self.ops.append(name)

    def reverse_rotate(self, s, name):
        if s:
            s.appendleft(s.pop())
            self.ops.append(name)

    def ra(self): self.rotate(self.a, "ra")
    def rb(self): self.rotate(self.b, "rb")
    def rra(self): self.reverse_rotate(self.a, "rra")
    def rrb(self): self.reverse_rotate(self.b, "rrb")
    def sa(self): self.swap(self.a, "sa")
    def sb(self): self.swap(self.b, "sb")
    def pa(self): self.push(self.b, self.a, "pa")
    def pb(self): self.push(self.a, self.b, "pb")

    def rr(self):
        if self.a and self.b:
            self.a.append(self.a.popleft())
            self.b.append(self.b.popleft())
            self.ops.append("rr")

    def rrr(self):
        if self.a and self.b:
            self.a.appendleft(self.a.pop())
            self.b.appendleft(self.b.pop())
            self.ops.append("rrr")

    # Helper to find optimal rotations for pushing element
    def get_target_position(self, value):
        # Find where value fits in stack a (sorted ascending)
        if not self.a:
            return 0
        positions = list(self.a)
        for i in range(len(positions)):
            prev = positions[i-1] if i > 0 else positions[-1]
            curr = positions[i]
            if prev < value <= curr:
                return i
        # If not found, position of smallest element
        min_val = min(positions)
        return positions.index(min_val)

    def rotate_to_top(self, stack, idx, stack_name):
        # Rotate stack so that element at idx is on top
        if idx <= len(stack)//2:
            for _ in range(idx):
                if stack_name == 'a':
                    self.ra()
                else:
                    self.rb()
        else:
            for _ in range(len(stack) - idx):
                if stack_name == 'a':
                    self.rra()
                else:
                    self.rrb()

    def rotate_both(self, a_rot, b_rot):
        # Rotate both stacks simultaneously for min moves
        while a_rot > 0 and b_rot > 0:
            self.rr()
            a_rot -= 1
            b_rot -= 1
        while a_rot > 0:
            self.ra()
            a_rot -= 1
        while b_rot > 0:
            self.rb()
            b_rot -= 1

        while a_rot < 0 and b_rot < 0:
            self.rrr()
            a_rot += 1
            b_rot += 1
        while a_rot < 0:
            self.rra()
            a_rot += 1
        while b_rot < 0:
            self.rrb()
            b_rot += 1

    def move_element(self):
        min_moves = None
        best = None
        for i, val in enumerate(self.b):
            target_pos = self.get_target_position(val)

            # Calculate rotations needed on a
            if target_pos <= len(self.a)//2:
                a_rot = target_pos
            else:
                a_rot = target_pos - len(self.a)

            # Calculate rotations needed on b
            if i <= len(self.b)//2:
                b_rot = i
            else:
                b_rot = i - len(self.b)

            # Calculate total moves (worst case sum)
            moves = max(abs(a_rot), abs(b_rot))
            if min_moves is None or moves < min_moves:
                min_moves = moves
                best = (i, val, a_rot, b_rot)

        i, val, a_rot, b_rot = best
        self.rotate_both(a_rot, b_rot)
        self.pa()

    def sort(self):
        # Index input
        values = list(self.a)
        indexed_values = [self.indexed[val] for val in values]
        self.a = deque(indexed_values)

        n = self.size
        chunk_size = 50 if n > 100 else 20

        i = 0
        # Push chunks to b
        while self.a:
            if self.a[0] <= i:
                self.pb()
                i += 1
            elif self.a[0] <= i + chunk_size:
                self.pb()
                self.rb()
                i += 1
            else:
                self.ra()

        # Now push all back in optimized way
        while self.b:
            self.move_element()

    def print_ops(self):
        return len(self.ops)

# Run test 100 times for 500 numbers
result = []
for _ in range(10):
    arr = random.sample(range(-1000, 1000), 500)
    stack = Stack(arr)
    stack.sort()
    result.append(stack.print_ops())

print(f"Min operations: {min(result)}")
print(f"Max operations: {max(result)}")
print(f"Average operations: {sum(result)/len(result):.2f}")
```
